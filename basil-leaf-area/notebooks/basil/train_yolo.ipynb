## Архивный ноутбук

В исследовании я использовал YOLO11-seg. В архивных ячейках я сохранил исходные имена весов и параметры; переносимый запуск обучения находится в `train.py`.


```
from ultralytics import YOLO

model = YOLO('yolov8m-seg.pt')  # load a pretrained model (recommended for training)

# Train the model
results = model.train(data='data/basil/data.yaml', epochs=25, imgsz=640, batch=8, patience=200)
```
